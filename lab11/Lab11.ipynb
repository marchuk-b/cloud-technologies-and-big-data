# Написати Docker Compose файл для двох сервісів
Ідея: познайомитись із інструментом Docker Compose як способом декларативно описувати та запускати мульти-контейнерні застосунки. <br>

Вимоги:
* Наявний файл compose.yaml у корені проєкту
* Описано два сервіси:
    * app — власний або готовий застосунок
    * redis або database — зовнішнє сховище
* Сервіс app має маппінг порту 8080:8080 (host:container)
* Визначена залежність сервісу app від redis чи database через depends_on
* Обидва сервіси мають бути в одній мережі (явно або неявно через Compose)
* Застосунок доступний на http://localhost:8080

Що протестувати:
* Compose файл валідний - `docker compose config`
* Усі сервіси запущені - `docker compose ps`
* Застосунок відповідає - `curl http://localhost:8080`
* Немає помилок у логах - `docker compose logs`
* Коректне завершення - `docker compose down`
