# Sharding (Сегментування/Шардінг)
Ідея: розподілити дані/навантаження по шардах за ключем.

## Вимоги
- Хеш-функція та стратегія:
    - Простий `mod N` або `consistent hashing` (кільце) — бажано надбудовою.

- Інтерфейс:
    - `put(key, value)`, `get(key)`, `moveShard(...)` (опціонально для ребалансування).
    - Візуалізація розподілу ключів по шардах (у логах/консолі).
    - Для демонстрації достатньо in-memory реалізації з кількох «шард-нод».

- Тести:
    - Рівномірність розподілу (для випадкових ключів) — статистично перевірити відхилення.
    - Стабільність при додаванні/видаленні шарда (для `consistent hashing`) — частка перемаплених ключів суттєво < 100%.
    - Коректність `get` після `put`.
