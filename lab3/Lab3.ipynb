# Retry
Ідея: безпечно повторювати операцію, яка може тимчасово падати.

## Вимоги

* Параметри: maxAttempts, стратегія паузи (constant, exponential, exponentialJitter), retryOn (перелік помилок/кодів).
* Метод: execute(fn) — викликає fn до успіху або вичерпання спроб.

Тести:
* Успіх з 1-ї спроби й з n-тої.
* Стоп після maxAttempts з правильною сумою затримок.
* Не повторюємо на non-retryable помилках.
* Мок fn, що падає певну кількість разів.
