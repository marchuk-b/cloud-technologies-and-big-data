# Throttle (Дросель)
Ідея: обмежити частоту виконання.

## Вимоги

* Параметри: rate (викликів/інтервал), або «token bucket» із capacity, refillRate.
* Метод: throttle(fn) → функція-обгортка, що гарантує не більше N викликів.
* Опції: leading, trailing, черга або відсікання надлишку.

Тести:
* Віконне обмеження: при 10 викликах за 1с і ліміті 3/с — виконується лише дозволена кількість.
* Режими leading/trailing.
* Черга: порядок збережено; відсікання: зайві відкинуто.
