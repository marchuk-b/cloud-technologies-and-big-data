# Debounce
Ідея: відкласти виконання до моменту, коли події припиняться.

## Вимоги

Параметри: delayMs.
Методи:
* debounce(fn) → повертає функцію-обгортку, яка викликає fn лише якщо події не було протягом delayMs.
* Опції: leading (виконати одразу перший), trailing (виконати останній).

Тести:
* Серія швидких викликів → єдиний виклик fn
* leading=true → миттєвий перший виклик + придушення наступних до паузи.
* Комбінації leading/trailing.
* Скасування таймера при dispose.
