# Запустити перший Docker-контейнер
Ідея: познайомитись із базовим робочим процесом Docker: встановити середовище виконання контейнерів, завантажити готовий образ із публічного реєстру та запустити перший контейнер <br>

Вимоги:
* Встановлений Rancher Desktop або Portainer
* Використати готовий образ — наприклад, nginxdemos/hello або containous/whoami
* Запустити контейнер із маппінгом порту 8080:80 (host:container)
* Контейнеру має бути присвоєне ім'я

Що протестувати:
* Контейнер запущений - `docker ps`
* Образ завантажено - `docker images`
* Застосунок відповідає - `curl http://localhost:8080`
* Логи контейнера - `docker logs <name>`
* Зупинка контейнера - `docker stop <name>`
* Видалення контейнера - `docker rm <name>`
