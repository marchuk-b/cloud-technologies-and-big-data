# Future (Майбутнє значення/Promise)
Ідея: представити результат, що стане доступним пізніше.

## Вимоги
- Інтерфейс:
    - `Future<T>` з методами `then`, `catch/onError`, `await/get(timeout)`.
    - Створення: `submit(fn)` -> `Future<T>`.
    - Комбінатори (мінімум один): `all([Future])` або `race([Future])`.
- Один раз встановлюється у `fulfilled` або `rejected`.
- Тести:
    - Успішне виконання та отримання значення.
    - Пропагація помилок.
    - `all` чекає всі; `race` — перший завершений.
    - Тайм-аут `get(timeout)`.
