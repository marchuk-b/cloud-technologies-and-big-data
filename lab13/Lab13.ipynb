# Написати Kubernetes маніфести Deployment та Service
Ідея: описати розгортання застосунку декларативно через YAML-маніфести. <br>

Вимоги:
- Структура файлів:
    - Маніфести розміщені в окремій директорії, наприклад `k3s/`
    - Два файли: `deployment.yaml` та `service.yaml`
- deployment.yaml
    - `apiVersion`: `apps/v1`, `kind: Deployment`
    - Вказано namespace (наприклад, `default` або власний)
    - Мінімум 2 репліки (`replicas: 2`)
    - Правильно налаштований selector та labels (мають збігатись)
    - У `spec.template.spec.containers`:
        - Власний образ із DockerHub (`image:`)
        - Вказано `containerPort`
- service.yaml
    - `kind`: `Service` типу `NodePort`
    - `selector` збігається з labels Deployment
    - Правильно вказані `port`, `targetPort` та `nodePort`

Що протестувати:
- Маніфести валідні - `kubectl apply --dry-run=client -f k3s/` → без помилок
- Deployment створено - `kubectl get deployment <your name>` → READY 2/2
- Поди запущені - `kubectl get pods` → статус Running для всіх подів
- Service створено - `kubectl get service <your name>` → тип NodePort
- Застосунок відповідає - `curl http://localhost:<nodePort>`
- Деталі Deployment - `kubectl describe deployment <your name>`
- Видалення та відновлення - видалити один под → Kubernetes автоматично створює новий
