# Time-out (Тайм-аут)
Ідея: обмежити максимальний час очікування операції.

## Вимоги

* Метод: withTimeout(fn, timeoutMs) — виконує fn і відхиляє/перериває після timeoutMs.
* Підтримка відміни — опціонально.

Тести:
* Успішна операція до тайм-ауту.
* Перевищення — повертається помилка тайм-ауту, побічні ефекти не відбулися/відмінені.
* Комбінування з Retry (не множити загальний час очікування безконтрольно).
