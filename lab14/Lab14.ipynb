# Написати Kubernetes маніфести Deployment та Service
Ідея: перенести всю інфраструктуру свого проєкта у Helm-chart. <br>

Вимоги:
- Структура чарту. Чарт має відповідати стандартній структурі Helm
- `Chart.yaml`
    - Заповнені поля: `name`, `version`, `appVersion`, `description`
    - Тип чарту: `application`
- `values.yaml` - мають бути винесені всі змінювані значення
- `templates/deployment.yaml`
    - Використовує `{{ .Values.replicaCount }}` для кількості реплік
    - Використовує `{{ .Values.image.repository }}`:`{{ .Values.image.tag }}` для образу
    - Labels та selector через `{{ include "course-app.fullname" . }}`
- `templates/service.yaml`
    - Тип, порти та nodePort – через `.Values.service.*`
    - Selector через `{{ include "course-app.fullname" . }}`

Що протестувати:
- Структура чарту валідна – `helm template <your name> <your name>/` → валідний YAML
- Інсталяція без помилок – `helm install <your name> <your name>/`
- Застосунок відповідає – `curl http://localhost:8080`
- Видалення чарту – `helm uninstall course-app` → всі ресурси видалено
