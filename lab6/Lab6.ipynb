# Fan-In (Мультиплексор)
Ідея: зібрати події/дані з кількох джерел у єдиний потік.

## Вимоги

* Інтерфейс: <br>
    `fanIn(inputs: Stream<T>[]) -> Stream<T>` <br>
    Гарантувати неблокуюче злиття, коректне закриття вихідного стріму, якщо всі вхідні завершені.

* Обробка помилок: або окремий канал помилок, або обгортання `Result<T, E>`.

Тести:
* Злиття двох/трьох потоків у один, порядок елементів збережено відносно кожного потоку (але не між потоками).
* Завершення після завершення всіх входів.
* Помилки одного входу не блокують інші (за обраною політикою).
