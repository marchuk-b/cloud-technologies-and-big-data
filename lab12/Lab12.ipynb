# Написати перевірку стану сервісів (Healthcheck)
Ідея: навчитись додавати healthcheck до сервісів у compose.yaml, щоб Docker міг автоматично визначати, чи сервіс справді готовий до роботи. <br>

Вимоги:
* Healthcheck додано до обох сервісів у compose.yaml
* Залежність app від redis/database оновлена до condition: service_healthy
* Після запуску обидва сервіси мають статус healthy (не просто running)

Що протестувати:
* Обидва сервіси healthy - docker compose ps → статус healthy
* Логи healthcheck-подій - docker compose logs — видно затримку старту app
