# Fan-Out (Демультиплексор)
Ідея: розподілити роботу з одного джерела на N воркерів.

## Вимоги
* Компоненти: черга задач, N воркерів, агрегатор результатів (опціонально).

* Інтерфейс: <br>
    `fanOut(process: (Task)->Result, workers: N) -> { submit(task), results() }` <br>

* Політика завершення: graceful shutdown.

Тести:
* Паралельність: час виконання з N воркерами менший, ніж з 1 (для штучно повільних задач).
* Безпека даних: відсутність гонок/подвійної обробки.
* Правильне завершення після no more tasks.
